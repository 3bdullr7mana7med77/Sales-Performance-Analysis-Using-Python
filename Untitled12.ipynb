```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)

products = [
    "Laptop Stand",
    "Wireless Mouse",
    "Keyboard",
    "Headset",
    "USB Hub",
    "Webcam"
]

regions = [
    "Cairo",
    "Alexandria",
    "Giza",
    "Mansoura",
    "Tanta"
]

n = 150

data = {
    "Order_ID": [f"ORD-{1000 + i}" for i in range(n)],

    "Date": pd.date_range(
        start="2026-01-01",
        periods=n,
        freq="D"
    ),

    "Product": np.random.choice(
        products,
        size=n
    ),

    "Region": np.random.choice(
        regions,
        size=n
    ),

    "Quantity": np.random.randint(
        1,
        10,
        size=n
    ),

    "Unit_Price": np.random.choice(
        [15, 20, 25, 30, 40, 50, 60, 75, 100],
        size=n
    )
}

df = pd.DataFrame(data)


df["Revenue"] = (
    df["Quantity"] *
    df["Unit_Price"]
)

# Cost is approximately 70% of revenue
df["Cost"] = (
    df["Revenue"] *
    np.random.uniform(
        0.55,
        0.80,
        size=n
    )
).round(2)

# Profit
df["Profit"] = (
    df["Revenue"] -
    df["Cost"]
).round(2)


df.loc[5, "Region"] = np.nan
df.loc[20, "Region"] = np.nan
df.loc[35, "Unit_Price"] = np.nan
df.loc[70, "Unit_Price"] = np.nan


print("=" * 60)
print("DATASET PREVIEW")
print("=" * 60)

print(df.head())

print("\n")

print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print(df.info())

print("\n")

print("=" * 60)
print("MISSING VALUES BEFORE CLEANING")
print("=" * 60)

print(df.isnull().sum())


# Convert Date column
df["Date"] = pd.to_datetime(df["Date"])

# Fill missing regions
df["Region"] = df["Region"].fillna("Unknown")

# Fill missing Unit_Price using median
df["Unit_Price"] = df["Unit_Price"].fillna(
    df["Unit_Price"].median()
)

# Recalculate Revenue
df["Revenue"] = (
    df["Quantity"] *
    df["Unit_Price"]
)

# Recalculate Profit
df["Profit"] = (
    df["Revenue"] -
    df["Cost"]
)

# Create Month column
df["Month"] = (
    df["Date"]
    .dt.to_period("M")
    .astype(str)
)



print("\n")
print("=" * 60)
print("MISSING VALUES AFTER CLEANING")
print("=" * 60)

print(df.isnull().sum())


total_revenue = df["Revenue"].sum()

total_cost = df["Cost"].sum()

total_profit = df["Profit"].sum()

total_orders = df["Order_ID"].nunique()

total_products = df["Product"].nunique()

average_order_value = df["Revenue"].mean()

profit_margin = (
    total_profit /
    total_revenue
) * 100

print("\n")
print("=" * 60)
print("KEY PERFORMANCE INDICATORS")
print("=" * 60)

print(f"Total Revenue       : {total_revenue:,.2f}")

print(f"Total Cost          : {total_cost:,.2f}")

print(f"Total Profit        : {total_profit:,.2f}")

print(f"Total Orders        : {total_orders}")

print(f"Total Products      : {total_products}")

print(
    f"Average Order Value : "
    f"{average_order_value:,.2f}"
)

print(
    f"Profit Margin       : "
    f"{profit_margin:.2f}%"
)


product_revenue = (
    df.groupby("Product")["Revenue"]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\n")
print("=" * 60)
print("REVENUE BY PRODUCT")
print("=" * 60)

print(product_revenue)


best_product = product_revenue.idxmax()

best_product_revenue = product_revenue.max()

print("\nBest Product:")
print(best_product)

print(
    f"Revenue: "
    f"{best_product_revenue:,.2f}"
)


plt.figure(figsize=(10, 5))

product_revenue.plot(
    kind="bar"
)

plt.title(
    "Revenue by Product",
    fontsize=16
)

plt.xlabel(
    "Product"
)

plt.ylabel(
    "Revenue"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()



region_revenue = (
    df.groupby("Region")["Revenue"]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\n")
print("=" * 60)
print("REVENUE BY REGION")
print("=" * 60)

print(region_revenue)



best_region = region_revenue.idxmax()

best_region_revenue = region_revenue.max()

print("\nBest Region:")
print(best_region)

print(
    f"Revenue: "
    f"{best_region_revenue:,.2f}"
)


plt.figure(figsize=(9, 5))

region_revenue.plot(
    kind="bar"
)

plt.title(
    "Revenue by Region",
    fontsize=16
)

plt.xlabel(
    "Region"
)

plt.ylabel(
    "Revenue"
)

plt.xticks(
    rotation=30
)

plt.tight_layout()

plt.show()



monthly_revenue = (
    df.groupby("Month")["Revenue"]
    .sum()
)

print("\n")
print("=" * 60)
print("MONTHLY REVENUE")
print("=" * 60)

print(monthly_revenue)



plt.figure(figsize=(11, 5))

monthly_revenue.plot(
    kind="line",
    marker="o"
)

plt.title(
    "Monthly Revenue",
    fontsize=16
)

plt.xlabel(
    "Month"
)

plt.ylabel(
    "Revenue"
)

plt.xticks(
    rotation=45
)

plt.grid(
    True
)

plt.tight_layout()

plt.show()


product_profit = (
    df.groupby("Product")["Profit"]
    .sum()
    .sort_values(
        ascending=False
    )
)

print("\n")
print("=" * 60)
print("PROFIT BY PRODUCT")
print("=" * 60)

print(product_profit)


plt.figure(figsize=(10, 5))

product_profit.plot(
    kind="bar"
)

plt.title(
    "Profit by Product",
    fontsize=16
)

plt.xlabel(
    "Product"
)

plt.ylabel(
    "Profit"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


top_orders = (
    df.sort_values(
        by="Revenue",
        ascending=False
    )
    .head(5)
)

print("\n")
print("=" * 60)
print("TOP 5 ORDERS")
print("=" * 60)

print(
    top_orders[
        [
            "Order_ID",
            "Product",
            "Region",
            "Quantity",
            "Revenue",
            "Profit"
        ]
    ]
)


print("\n")
print("=" * 60)
print("BUSINESS INSIGHTS")
print("=" * 60)

print(
    f"1. Total revenue is "
    f"{total_revenue:,.2f}."
)

print(
    f"2. Total profit is "
    f"{total_profit:,.2f}."
)

print(
    f"3. The best-performing product is "
    f"{best_product}."
)

print(
    f"4. The best-performing region is "
    f"{best_region}."
)

print(
    f"5. Average order value is "
    f"{average_order_value:,.2f}."
)

print(
    f"6. Overall profit margin is "
    f"{profit_margin:.2f}%."
)

print("\n")
print("=" * 60)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 60)
```
